```python
# Intro to Euler's Method

## The Idea on how it works 
#  For an ODE dy/dt = f(y,t) (Ordinary Deffrential Equation)
#  dy/dt or rather f(y,t) is the slope of the current point 
#  We take a small time step h 
#  And for that small step we move along the slope : y_new = y_old + h*f(y_old, t_old) 
# 
#  This is Euler's Method and it works because Euler's Theorem keeps only the first 2 from the Taylor Expansion: 
#  y(t+h) = y(t) + h*y'(t) + (h^2/2)*y''(t) .....
```
